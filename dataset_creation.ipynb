```python
associated_query_ids = []

for _, row in temp_df.iterrows():
    associated_query = query_df.loc[
        (query_df["uid"]==row["uid"]) 
        & (query_df["topic_number"]==row["topic_number"]) 
        & (query_df["query"]==row["query"]) 
        & (query_df["timestamp"]<row["timestamp"])
    ]

    if len(associated_query) == 0:
        print("No query found for serp id: ", row["serp_id"], "; uid: ", row["uid"], "; query: ", row["query"], "; time: ", row["timestamp"])
        # associated_query = query_df.loc[
        #     (query_df["uid"]==row["uid"]) 
        #     & (query_df["topic_number"]==row["topic_number"]) 
        #     # & (query_df["query"]==row["query"]) 
        #     & (query_df["timestamp"]<row["timestamp"])
        # ]
        # print("Matching based on uid and topic number (w/o query match) led to ", len(associated_query), " matches.\n")
        associated_query_ids.append(None)

    elif len(associated_query) > 1:
        # print("multiple queries found: ", row["serp_id"], row["uid"], row["timestamp"])
        time_diff = np.inf
        for _, query in associated_query.iterrows():
            # print(query)
            time_diff_compute = row["timestamp"] - query["timestamp"]
            if time_diff >= time_diff_compute.total_seconds():
                correct_query = query
                time_diff = time_diff_compute.total_seconds()
        # print("query chosen for ", row["serp_id"], "query: ", query["query"], "; timestamp: ", query["timestamp"])
        associated_query_ids.append(correct_query["qid"])
    else:
        associated_query_ids.append(associated_query["qid"].values[0])

temp_df["qid"] = associated_query_ids
```

```python
import pandas as pd

serp_df = pd.read_csv("Data/logs_compiled.csv")
relevance = pd.read_csv("Data/relevance.csv")
relevance_score = []

for _, row in serp_df.iterrows():
    relevant_docs = relevance.loc[relevance["topic_number"]==row["topic_number"]]["doc"].tolist()
    if row["url"] in relevant_docs:
        relevance_score.append(1)
    else:
        relevance_score.append(0)

serp_df["relevance_score"] = relevance_score
serp_df.to_csv("Data/logs_compiled.csv", index=False, encoding="utf8")
```

```python
temp = serp_df.loc[(serp_df["rank"]<=10)|(pd.isna(serp_df["rank"]))].groupby(["serp_id"])["relevance_score"].sum()/10
df = pd.DataFrame(temp).reset_index()
df.columns = ["serp_id", "precision@10"]
df["query"] = [serp_df.loc[serp_df["serp_id"]==serp_id]["query"].values[0] for serp_id in df["serp_id"]]
df["topic_number"] = [serp_df.loc[serp_df["serp_id"]==serp_id]["topic_number"].values[0] for serp_id in df["serp_id"]]
df["uid"] = [serp_df.loc[serp_df["serp_id"]==serp_id]["uid"].values[0] for serp_id in df["serp_id"]]
df = df[["uid", "topic_number", "serp_id", "query", "precision@10"]]
df = df.sort_values(by=["uid"]).reset_index(drop=True)
df = df[["uid","topic_number","query","serp_id", "precision@10"]]
df.to_csv("Data/query_formulation_df.csv", index=False, encoding='utf8')
```

### Navigation quality

```python
import pandas as pd

navigation_quality = []
results_navigation = pd.read_csv("Data/results_navigation.csv")
results_navigation["timestamp"] = pd.to_datetime(results_navigation["timestamp"])

click_df = pd.read_csv("Data/clicked_results_df.csv")
click_df["timestamp"] = pd.to_datetime(click_df["timestamp"])

for serp_id in results_navigation["serp_id"].unique():
    serp = results_navigation.loc[results_navigation["serp_id"]==serp_id]
    clicked_results = click_df.loc[click_df["serp_id"]==serp_id]

    uid = serp["uid"].values[0]
    topic_number = serp["topic_number"].values[0]
    qid = serp["qid"].values[0]
    query = serp["query"].values[0]
    total_clicks = serp["click_count"].sum()

    if total_clicks == 0:
        click_accuracy = None
        if sum(serp["relevance_score"]) == 0: # no relevant resources in the SERP
            effective_navigation = 1
        elif sum(serp["relevance_score"]) > 0: # there are relevant resources in the SERP
            effective_navigation = 0
    else:
        clicked_resources = 0
        positive_clicks = 0
        for _, result in serp.loc[serp["click_count"]>0].iterrows(): # serp results that were clicked
            clicked_resources += 1
            if result["relevance_score"] == 1:
                positive_clicks += 1
        click_accuracy = round(positive_clicks/clicked_resources,2)
        effective_navigation = int(click_accuracy >= 0.80)

    repetitions = total_clicks - len(clicked_results["rank"].unique())
    last_relevant_timestamp = serp["timestamp"].values[0]
    found_a_relevant_click = 0
    for _, click in clicked_results.iterrows():
        relevance = serp.loc[(serp["url"]==click["url"]) & (serp["rank"]==click["rank"])]["relevance_score"].values[0]
        if relevance == 1:
            found_a_relevant_click = 1
            last_relevant_timestamp = click["timestamp"]

    tail = len(clicked_results.loc[clicked_results["timestamp"]>last_relevant_timestamp])

    if found_a_relevant_click == 0:
            last_relevant_timestamp = None

    # if total_clicks > 0:
    #     nav_eff = (total_clicks - repetitions - tail)/total_clicks
    # else:
    #      nav_eff = None

    navigation_quality.append([uid, topic_number, qid, query, serp_id, total_clicks, repetitions, tail, last_relevant_timestamp, click_accuracy, effective_navigation])

navigation_quality_df = pd.DataFrame(navigation_quality, columns = ["uid", "topic_number", "qid", "query", "serp_id", "total_clicks", "repetitions", "tail", "last_relevant_timestamp", "click_accuracy", "effective_navigation"])
navigation_quality_df.to_csv("Data/navigation_quality.csv", index=False, encoding="utf8")
navigation_quality_df.head(10)
```

```python
click_df = pd.read_csv("Data/clicked_results_df.csv")
click_df["timestamp"] = pd.to_datetime(click_df["timestamp"])

part_answer_df = pd.read_csv("Data/part_answers_df.csv")
part_answer_df["timestamp"] = pd.to_datetime(part_answer_df["timestamp"])
part_answer_df["associated_click_time"] = pd.to_datetime(part_answer_df["associated_click_time"])

relevance_df = pd.read_csv("Data/relevance.csv")

answer_list = []
answer_time_list = []
relevance = []
for _, row in click_df.iterrows():
    click_time = row["timestamp"]
    topic_num = row["topic_number"]
    relevant_urls = relevance_df.loc[relevance_df["topic_number"]==topic_num]["doc"].tolist()
    url = row["url"]
    relevance.append(int(url in relevant_urls))
    part_answers = part_answer_df.loc[part_answer_df["associated_click_time"]==click_time]
    if len(part_answers) == 0: # user did not interact with answer box after this click
        answer = None
        answer_time = None
    else:
        answer_time = part_answers["timestamp"].max() # taking the last instance of opening the answer box to enter an answer after a click
        answer = part_answers.loc[part_answers["timestamp"]==answer_time]["answer"].values[0]
    answer_list.append(answer)
    answer_time_list.append(answer_time)

click_df["relevance_score"] = relevance
click_df["answer"] = answer_list
click_df["answer_time"] = answer_time_list
```

```python
import spacy

nlp = spacy.load("it_core_news_lg")

ref_ans = []
change_from_prev_answer = []

for uid in click_df["uid"].unique():
    user_sessions = click_df.loc[click_df["uid"]==uid]
    for topic_num in user_sessions["topic_number"].unique():
        prev_ans = None # start with no previous answer for a task
        clicks = user_sessions.loc[user_sessions["topic_number"]==topic_num].reset_index()
        for i, row in clicks.iterrows():
            current_ans = row["answer"] # the answer entered after a click
            if pd.isna(current_ans): # answer box is empty
                current_ans = ""
            current_ans = current_ans.lower()
            current_ans_tokens = [token.text for token in nlp(current_ans)]
            if i == 0: # first time opening the answer box for this task
                ans_diff = current_ans_tokens
            else:
                if pd.isna(prev_ans):
                    prev_ans = ""
                prev_ans = prev_ans.lower()
                prev_ans_tokens = [token.text for token in nlp(prev_ans)]
                ans_diff = list(set(current_ans_tokens).difference(set(prev_ans_tokens)))

            ref_ans.append(prev_ans)
            change_from_prev_answer.append(ans_diff)
            if pd.isna(row["answer_time"])==False: # if the user did open the answer box after current click
                prev_ans = current_ans # answered entered into the answer box after the current click becomes the reference answer for subsequent clicks

click_df["answer_previously"] = ref_ans
click_df["change_from_prev_ans"] = change_from_prev_answer
```

```python
effective_distillation = []
problems = {}

for uid in click_df["uid"].unique():
    user_clicks = click_df.loc[click_df["uid"]==uid]
    for topic_num in user_clicks["topic_number"].unique():
        session_clicks = user_clicks.loc[user_clicks["topic_number"]==topic_num].reset_index()
        session_clicks["answer"] = session_clicks["answer"].fillna(value="")
        session_clicks["answer_previously"] = session_clicks["answer_previously"].fillna(value="")
        # print(session_clicks)
        for i, row in session_clicks.iterrows():
            if row["relevance_score"]==0: # clicked on an irrelevant result
                if pd.isna(row["answer_time"]): # did not open the answer box
                    effective_distillation.append(1)
                else: # opened the answer box
                    if row["answer"].lower()==row["answer_previously"].lower(): # answer did not change
                        effective_distillation.append(1)
                    else:
                        effective_distillation.append(0) # changed the answer
            elif row["relevance_score"]==1: # clicked on a relevant result
                if i == 0: # first answering attempt
                    if row["answer_match_ratio"] > 0: # added at least one correct keyword
                        effective_distillation.append(1)
                    else: # did not add any correct keywords
                        effective_distillation.append(0)
                else: # subsequent answering attempts
                    if row["answer"] == row["answer_previously"]: # answer did not change
                        if row["answer_match_ratio"] == 1: # already have full correct answer
                            effective_distillation.append(1)
                        else: # answer is not fully correct yet
                            effective_distillation.append(0)
                    else: # changed answer
                        prev_row = session_clicks.iloc[i-1]
                        if row["answer_match_ratio"] > prev_row["answer_match_ratio"]: # improved their answer
                            effective_distillation.append(1)
                        else: # answer did not improve
                            effective_distillation.append(0)
            else:
                print("Result is neither relevant nor irrelevant??")

# print(effective_distillation)
click_df["effective_distillation"] = effective_distillation
```

## Final answers

```python
import pandas as pd
import json
# import spacy

# nlp = spacy.load("it_core_news_lg")

final_answer_df = pd.read_csv("Data/final_answer_df.csv")
final_answer_df["timestamp"] = pd.to_datetime(final_answer_df["timestamp"])
final_answer_df = final_answer_df[["uid", "topic_number", "prompt", "answer", "TaskStartOrEnd", "timestamp"]]

final_answers = []

for _, row in final_answer_df[final_answer_df["TaskStartOrEnd"]=="TaskEnd"].iterrows():
    uid = row["uid"]
    topic_num = row["topic_number"]
    start_time = final_answer_df.loc[(final_answer_df["uid"]==uid) & (final_answer_df["topic_number"]==topic_num) & (final_answer_df["TaskStartOrEnd"]=="TaskStart")]["timestamp"].values[0]
    task_duration = (row["timestamp"] - start_time).total_seconds()/60
    final_answers.append([uid, topic_num, row["prompt"], row["answer"], start_time, row["timestamp"], task_duration])

final_answer_df = pd.DataFrame(final_answers, columns=["uid", "topic_number", "prompt", "answer", "TaskStartTime", "TaskEndTime", "TaskDuration"])
# final_answer_df

with open("Data/answer_set.json", 'r') as f:
        answer_set = json.load(f)

reqd_keywords_map = {
        "g3t1": 1,
        "g3t2": 2,
        "g3t3": 5,
        "g4t1": 4,
        "g4t2": 2,
        "g4t3": 3
}

match_counts = []
matched_keywords = []
for _, row in final_answer_df.iterrows():
        num_matches = 0
        answers = answer_set[row["topic_number"]]
        matched_answers = []
        for answer_key in answers.keys():
                if answer_key.title() in row["answer"].title():
                        matched_answers.append(answer_key)
                        num_matches += 1
                else:
                        for answer in answers[answer_key]:
                                if answer.title() in row["answer"].title():
                                # print(row["answer"], ";", answer)
                                        matched_answers.append(answer)
                                        num_matches += 1
                                        break
        # print(row["uid"], row["topic_number"], row["answer"], num_matches)
        match_counts.append(num_matches)
        matched_keywords.append(matched_answers)

final_answer_df["num_answer_matches"] = match_counts
final_answer_df["matched_keywords"] = matched_keywords
final_answer_df["answer_match_ratio"] = [round(min(row["num_answer_matches"]/reqd_keywords_map[row["topic_number"]],1.00),2) for _, row in final_answer_df.iterrows()]
```
