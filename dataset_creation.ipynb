```python
import json
import pandas as pd
import glob
import numpy as np
import driver
```

```python
query_df = pd.DataFrame()
query_actions_df = pd.DataFrame() 
serp_df = pd.DataFrame()
clicked_results_df = pd.DataFrame()
viewed_results_df = pd.DataFrame()
final_answer_df = pd.DataFrame()

for file_name in glob.glob("Data/3rd/*_FULL.log"):
    # print(file_name)
    with open(file_name, 'r') as f:
        result = [json.loads(x) for x in f.readlines()]
    
    # query actions
    query_temp_df = driver.get_queries(result)
    query_df = pd.concat([query_df, query_temp_df], ignore_index=True)

    # query actions
    query_actions_temp_df = driver.get_query_actions(result)
    query_actions_df = pd.concat([query_actions_df, query_actions_temp_df], ignore_index=True)

    # SERPs generated
    serp_temp_df = driver.get_generated_serps(result)
    serp_df = pd.concat([serp_df, serp_temp_df], ignore_index=True)

    # Clicked results
    clicked_results_temp_df = driver.get_clicked_results(result)
    clicked_results_df = pd.concat([clicked_results_df, clicked_results_temp_df], ignore_index=True)

    # Viewed results
    viewed_results_temp_df = driver.get_viewed_results(result) 
    viewed_results_df = pd.concat([viewed_results_df, viewed_results_temp_df])

    # final answers
    final_answer_temp_df = driver.get_final_answers(result)
    final_answer_df = pd.concat([final_answer_df, final_answer_temp_df], ignore_index=True)

for file_name in glob.glob("Data/4th/*_FULL.log"):
    # print(file_name)
    with open(file_name, 'r') as f:
        result = [json.loads(x) for x in f.readlines()]
    
    # query actions
    query_temp_df = driver.get_queries(result)
    query_df = pd.concat([query_df, query_temp_df], ignore_index=True)

    # query actions
    query_actions_temp_df = driver.get_query_actions(result)
    query_actions_df = pd.concat([query_actions_df, query_actions_temp_df], ignore_index=True)

    # SERPs generated
    serp_temp_df = driver.get_generated_serps(result)
    serp_df = pd.concat([serp_df, serp_temp_df], ignore_index=True)

    # Clicked results
    clicked_results_temp_df = driver.get_clicked_results(result)
    clicked_results_df = pd.concat([clicked_results_df, clicked_results_temp_df], ignore_index=True)

    # Viewed results
    viewed_results_temp_df = driver.get_viewed_results(result) 
    viewed_results_df = pd.concat([viewed_results_df, viewed_results_temp_df])

    # final answers
    final_answer_temp_df = driver.get_final_answers(result)
    final_answer_df = pd.concat([final_answer_df, final_answer_temp_df], ignore_index=True)

## reordering all dataframes uid, topic_number, and timestamp
query_df = query_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)
query_actions_df = query_actions_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)
serp_df = serp_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)
clicked_results_df = clicked_results_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)
viewed_results_df = viewed_results_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)
final_answer_df = final_answer_df.sort_values(["uid", "timestamp"]).reset_index(drop=True)


## assigning qid
query_df["qid"] = ["q"+str(i) for i in range(1, len(query_df)+1)]
query_df = query_df[["uid", "topic_number", "qid", "query", "source", "timestamp"]]


## assigning serp id
serp_num = 0
serp_id_list = []

for _, row in serp_df.iterrows():
    if row["rank"]==1 or pd.isna(row["rank"]):
        serp_num += 1
    serp_id_list.append("serp"+str(serp_num))

serp_df["serp_id"] = serp_id_list
serp_df = serp_df[["uid", "topic_number", "query", "serp_id", "url", "title", "snippet", "rank", "timestamp"]]

## removing duplicates
final_answer_df = final_answer_df.drop_duplicates(ignore_index=True)
```

```python
import pandas as pd
import json
# import spacy

# nlp = spacy.load("it_core_news_sm")

final_answer_df = pd.read_csv("Data/final_answer_df.csv")

with open("Data/answer_set.json", 'r') as f:
        answer_set = json.load(f)

match_counts = []
matched_keywords = []
for _, row in final_answer_df.iterrows():
        num_matches = 0
        answers = answer_set[row["topic_number"]]
        matched_answers = []
        for answer_key in answers.keys():
                for answer in answers[answer_key]:
                        if answer.title() in row["answer"].title():
                        # print(row["answer"], ";", answer)
                                matched_answers.append(answer)
                                num_matches += 1
                                break
        print(row["uid"], row["topic_number"], row["answer"], num_matches)
        match_counts.append(num_matches)
        matched_keywords.append(matched_answers)

final_answer_df["num_answer_matches"] = match_counts
final_answer_df["matched_keywords"] = matched_keywords
```
